## Tujuan

Chapter ini membahas Exploratory Data Analysis (EDA), yaitu proses untuk memahami
karakteristik data sebelum analisis lanjutan atau pemodelan.

Topik yang dibahas mengikuti struktur chapter:
- structured data dan rectangular data
- estimates of location
- estimates of variability
- exploring data distribution
- binary dan categorical data
- correlation
- exploring multiple variables

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
```

## 1. Structured Data dan Rectangular Data

Data berbentuk tabel dapat dipandang sebagai kumpulan observasi pada baris dan variabel pada kolom.

```python
n = 1000

df = pd.DataFrame({
    "weight": np.random.normal(68, 8, n),
    "height": np.random.normal(170, 7, n),
    "age": np.random.randint(18, 60, n),
    "score": np.random.normal(75, 10, n),
    "gender": np.random.choice(["Female", "Male"], n),
    "category": np.random.choice(["A", "B", "C"], n, p=[0.5, 0.3, 0.2])
})

df.head()
```

```python
print("Shape:", df.shape)
print("\nTipe data:")
print(df.dtypes)
print("\nMissing value:")
print(df.isna().sum())
```

## 2. Estimates of Location

Mean, median, weighted mean, dan trimmed mean digunakan untuk menggambarkan lokasi/pusat data.

```python
from scipy.stats import trim_mean

x = df["score"]

mean_value = x.mean()
median_value = x.median()
trimmed_value = trim_mean(x, proportiontocut=0.1)

print("Mean:", mean_value)
print("Median:", median_value)
print("Trimmed mean:", trimmed_value)
```

### Weighted Mean

Weighted mean memberi kontribusi yang berbeda pada setiap observasi berdasarkan bobot.

```python
weights = np.random.uniform(0.5, 2.0, size=len(df))

weighted_mean = np.average(
    df["score"],
    weights=weights
)

print("Weighted mean:", weighted_mean)
```

## 3. Estimates of Variability

Variabilitas menunjukkan seberapa menyebar data dari pusat distribusinya.

```python
variance = x.var()
std = x.std()
data_range = x.max() - x.min()
q1 = x.quantile(0.25)
q3 = x.quantile(0.75)
iqr = q3 - q1

summary = pd.Series({
    "Variance": variance,
    "Standard deviation": std,
    "Range": data_range,
    "Q1": q1,
    "Q3": q3,
    "IQR": iqr
})

summary
```

## 4. Exploring Data Distribution

Histogram membantu melihat frekuensi pada rentang nilai. Density plot memberikan gambaran bentuk distribusi secara halus.

```python
plt.figure(figsize=(7, 4))
plt.hist(x, bins=30)
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title("Distribution of Score")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.kdeplot(x, fill=False)
plt.xlabel("Score")
plt.ylabel("Density")
plt.title("Density Plot of Score")
plt.show()
```

## 5. Boxplot dan Outlier

Boxplot menampilkan median, quartile, dan observasi yang berada jauh dari pola utama.

```python
plt.figure(figsize=(7, 3))
sns.boxplot(x=x)
plt.title("Boxplot of Score")
plt.show()
```

## 6. Percentile dan Quantile

Percentile digunakan untuk menentukan posisi relatif suatu observasi dalam distribusi.

```python
percentiles = {
    "10%": x.quantile(0.10),
    "25%": x.quantile(0.25),
    "50%": x.quantile(0.50),
    "75%": x.quantile(0.75),
    "90%": x.quantile(0.90),
}

print(percentiles)
```

## 7. Binary dan Categorical Data

Variabel kategorikal menyimpan label atau kelompok. Analisis awal dapat dilakukan melalui frekuensi masing-masing kategori.

```python
print(df["gender"].value_counts())
print()
print(df["category"].value_counts())
```

```python
df["category"].value_counts().plot(kind="bar")
plt.xlabel("Category")
plt.ylabel("Count")
plt.title("Category Frequency")
plt.show()
```

## 8. Correlation

Correlation mengukur hubungan linear antara dua variabel numerik. Nilainya berada antara -1 dan 1. Correlation bukan bukti sebab-akibat.

```python
corr = df[["weight", "height", "age", "score"]].corr()
corr
```

```python
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()
```

## 9. Scatterplot

Scatterplot digunakan untuk melihat pola hubungan antara dua variabel numerik.

```python
plt.figure(figsize=(6, 4))
plt.scatter(df["height"], df["weight"], alpha=0.35)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

## 10. Exploring Multiple Variables

Saat beberapa variabel dianalisis bersama, kombinasi statistik dan visualisasi membantu menemukan pola, hubungan, dan potensi anomali.

```python
sns.pairplot(df[["weight", "height", "age", "score"]].sample(250, random_state=42))
plt.show()
```

## Interpretasi

Dari EDA, lokasi data dapat dipahami melalui mean dan median, sedangkan
penyebaran data dapat dilihat melalui standard deviation, variance, range,
dan IQR. Histogram dan density plot membantu melihat bentuk distribusi,
boxplot membantu mengamati potensi outlier, sedangkan correlation dan
scatterplot membantu melihat hubungan antarvariabel.
