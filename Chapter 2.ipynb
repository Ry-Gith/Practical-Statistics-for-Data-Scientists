## Tujuan

Chapter ini membahas population, sample, sampling bias, sampling distribution,
Central Limit Theorem, standard error, bootstrap, confidence interval, serta
beberapa probability distribution.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm, t, binom, poisson, chi2, f, expon, weibull_min

np.random.seed(42)
```

## 1. Population dan Random Sampling

Population adalah keseluruhan objek yang ingin dipelajari. Sample adalah sebagian observasi yang dipilih dari population.

```python
population = np.arange(1, 10001)

sample = np.random.choice(
    population,
    size=100,
    replace=False
)

print("Population size:", len(population))
print("Sample size:", len(sample))
print("Sample mean:", sample.mean())
```

## 2. Sample Bias

Sample bias muncul ketika proses pemilihan sample membuat sample tidak lagi mewakili population dengan baik.

```python
population = np.random.normal(50, 10, 100000)

random_sample = np.random.choice(
    population,
    size=500,
    replace=False
)

biased_sample = population[population > 60][:500]

print("Population mean:", population.mean())
print("Random sample mean:", random_sample.mean())
print("Biased sample mean:", biased_sample.mean())
```

## 3. Sampling Distribution

Sampling distribution adalah distribusi suatu statistik yang diperoleh dari pengulangan proses sampling.

```python
population = np.random.normal(50, 10, 100000)

sample_means = []

for _ in range(2000):
    sample = np.random.choice(
        population,
        size=50,
        replace=True
    )
    sample_means.append(sample.mean())

sample_means = np.array(sample_means)

print("Mean of sample means:", sample_means.mean())
print("SD of sample means:", sample_means.std(ddof=1))
```

```python
plt.figure(figsize=(7, 4))
plt.hist(sample_means, bins=35)
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Sampling Distribution of the Mean")
plt.show()
```

## 4. Central Limit Theorem

Central Limit Theorem menjelaskan bahwa distribusi sample mean akan mendekati bentuk normal ketika ukuran sample cukup besar, walaupun population awal tidak normal.

```python
population = np.random.exponential(scale=10, size=100000)

means_small = []
means_large = []

for _ in range(1000):
    means_small.append(
        np.random.choice(population, size=5, replace=True).mean()
    )
    means_large.append(
        np.random.choice(population, size=50, replace=True).mean()
    )

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].hist(means_small, bins=30)
axes[0].set_title("Sample Size = 5")

axes[1].hist(means_large, bins=30)
axes[1].set_title("Sample Size = 50")

plt.tight_layout()
plt.show()
```

## 5. Standard Error

Standard error mengukur variasi statistik sample antar pengambilan sample. Untuk mean, standard error dapat diperkirakan dengan SD sample dibagi akar ukuran sample.

```python
sample = np.random.normal(50, 10, 100)

se = sample.std(ddof=1) / np.sqrt(len(sample))

print("Standard error:", se)
```

## 6. Bootstrap

Bootstrap melakukan resampling dengan replacement dari data yang tersedia untuk mendekati distribusi suatu statistic.

```python
data = np.array([10, 12, 15, 14, 11, 16, 18, 13, 17, 15])

bootstrap_means = []

for _ in range(5000):
    bootstrap_sample = np.random.choice(
        data,
        size=len(data),
        replace=True
    )
    bootstrap_means.append(bootstrap_sample.mean())

bootstrap_means = np.array(bootstrap_means)

print("Original mean:", data.mean())
print("Bootstrap mean:", bootstrap_means.mean())
```

```python
plt.figure(figsize=(7, 4))
plt.hist(bootstrap_means, bins=35)
plt.xlabel("Bootstrap Mean")
plt.ylabel("Frequency")
plt.title("Bootstrap Distribution")
plt.show()
```

## 7. Confidence Interval

Confidence interval memberikan rentang estimasi berdasarkan ketidakpastian sampling.

```python
lower = np.percentile(bootstrap_means, 2.5)
upper = np.percentile(bootstrap_means, 97.5)

print("95% bootstrap interval:", (lower, upper))
```

## 8. Normal Distribution

Normal distribution berbentuk simetris dan sering menjadi model dasar dalam analisis statistik.

```python
x = np.linspace(-4, 4, 500)

plt.figure(figsize=(7, 4))
plt.plot(x, norm.pdf(x))
plt.title("Standard Normal Distribution")
plt.xlabel("z")
plt.ylabel("Density")
plt.show()
```

## 9. t-Distribution

```python
x = np.linspace(-4, 4, 500)

plt.figure(figsize=(7, 4))
plt.plot(x, t.pdf(x, df=10))
plt.plot(x, norm.pdf(x))
plt.legend(["t distribution, df=10", "normal distribution"])
plt.title("t-Distribution vs Normal")
plt.show()
```

## 10. Binomial Distribution

```python
x = np.arange(0, 21)

prob = binom.pmf(
    x,
    n=20,
    p=0.5
)

plt.figure(figsize=(7, 4))
plt.bar(x, prob)
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

## 11. Chi-Square Distribution

```python
x = np.linspace(0, 20, 500)

plt.figure(figsize=(7, 4))
plt.plot(x, chi2.pdf(x, df=5))
plt.title("Chi-Square Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

## 12. F-Distribution

```python
x = np.linspace(0, 8, 500)

plt.figure(figsize=(7, 4))
plt.plot(x, f.pdf(x, dfn=5, dfd=10))
plt.title("F-Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

## 13. Poisson Distribution

```python
x = np.arange(0, 20)

prob = poisson.pmf(
    x,
    mu=5
)

plt.figure(figsize=(7, 4))
plt.bar(x, prob)
plt.xlabel("Number of events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

## 14. Exponential Distribution

```python
x = np.linspace(0, 10, 500)

plt.figure(figsize=(7, 4))
plt.plot(x, expon.pdf(x, scale=2))
plt.title("Exponential Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

## 15. Weibull Distribution

```python
x = np.linspace(0, 5, 500)

plt.figure(figsize=(7, 4))
plt.plot(x, weibull_min.pdf(x, c=1.8, scale=2))
plt.title("Weibull Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

## Interpretasi

Sampling membantu menggunakan sebagian data untuk mempelajari population.
Sampling distribution menunjukkan variasi suatu statistic ketika sampling
diulang. Central Limit Theorem menjelaskan bentuk distribusi sample mean pada
ukuran sample yang cukup besar. Bootstrap memberi pendekatan resampling untuk
mengestimasi ketidakpastian tanpa harus mengetahui distribusi population secara
persis.
